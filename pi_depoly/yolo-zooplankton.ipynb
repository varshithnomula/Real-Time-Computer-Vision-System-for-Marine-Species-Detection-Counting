```python
import yaml
import os

classes = sorted(classes)

data_config = {
    'train': f'{KAGGLE_DATA_DIR}/images/train',
    'val': f'{KAGGLE_DATA_DIR}/images/val',
    'nc': len(classes),
    'names': classes
}

# In Kaggle, we write to the /kaggle/working/ directory
with open('/kaggle/working/data.yaml', 'w') as f:
    yaml.dump(data_config, f)

print("Created data.yaml for Kaggle environment.")
```

```python
from ultralytics import YOLO

# Load a pretrained YOLOv8n model
model = YOLO('yolov8n.pt')

# Train the model
results = model.train(
    data='/kaggle/working/data.yaml',
    epochs=32,
    imgsz=640,
    batch=16,
    project='/kaggle/working/train_results',
    name='yolov8n_zooplankton',
    exist_ok=True
)
```
